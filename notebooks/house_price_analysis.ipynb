# 🏠 House Price Prediction — Exploratory Data Analysis (EDA)

This notebook conducts end-to-end Exploratory Data Analysis (EDA) on the house dataset to understand feature distributions, data hygiene, correlations, and key relationships with house prices.

## 1. Import Libraries & Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Load dataset
df = pd.read_csv("../data/house_data.csv")
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

## 2. Data Understanding & Profiling

We inspect missing values, data types, and duplicate records.

```python
print("=== DATASET INFORMATION ===")
print(df.info())

print("\n=== MISSING VALUES ===")
print(df.isnull().sum())

print("\n=== DUPLICATE ROWS ===")
print(f"Duplicates count: {df.duplicated().sum()}")

print("\n=== SUMMARY STATISTICS ===")
df.describe().T
```

## 3. Exploratory Visualizations

### Visualization 1: Distribution of House Prices

```python
plt.figure(figsize=(10, 5))
sns.histplot(df["price"], kde=True, color="#0284c7", bins=40)
plt.title("Distribution of House Prices ($)", fontsize=14, fontweight="bold")
plt.xlabel("Price ($)")
plt.ylabel("Count")
plt.show()

print(f"Median Price: ${df['price'].median():,.2f}")
print(f"Mean Price:   ${df['price'].mean():,.2f}")
```

### Visualization 2: Living Area (sqft_living) vs House Price

```python
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x="sqft_living", y="price", scatter_kws={"alpha": 0.4, "color": "#0284c7"}, line_kws={"color": "#dc2626"})
plt.title("Living Area (sqft) vs House Price ($)", fontsize=14, fontweight="bold")
plt.xlabel("Living Area (sq ft)")
plt.ylabel("Price ($)")
plt.show()
```

### Visualization 3: Correlation Matrix Heatmap

```python
numeric_df = df.select_dtypes(include=[np.number]).drop(columns=["id"], errors="ignore")
corr = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Feature Correlation Heatmap", fontsize=14, fontweight="bold")
plt.show()
```

### Visualization 4: Construction Grade vs House Price Boxplot

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="grade", y="price", palette="Blues")
plt.title("House Price Distribution by Construction Grade", fontsize=14, fontweight="bold")
plt.xlabel("Grade (1 = Low, 13 = High Luxury)")
plt.ylabel("Price ($)")
plt.show()
```

## 4. Key Findings & EDA Summary

1. **Right-Skewed Target Distribution:** House prices are right-skewed with most values under $1M, with a few high-end luxury properties.
2. **Strongest Predictors:** `sqft_living`, `grade`, and `sqft_above` show the highest positive correlation with price (r > 0.65).
3. **Construction Grade:** Property price grows exponentially with construction quality grade.
4. **Data Quality:** The dataset is exceptionally clean with minimal missing values and 0 duplicate rows.
